## Solution 1 — select, filter, withColumn

````python
q1 = (
    employeesDF
    .select("name", "department", "salary")
    .filter((F.col("department") == "Engineering") & (F.col("salary") > 90000))
    .withColumn("salary_inr", F.col("salary") * 83)
)

q1.show()
````

## Solution 2 — groupBy + agg

````python
q2 = (
    employeesDF
    .groupBy("department")
    .agg(
        F.count("emp_id").alias("total_employees"),
        F.round(F.avg("salary"), 2).alias("avg_salary"),
        F.max("salary").alias("max_salary"),
        F.min("salary").alias("min_salary"),
        F.sum("salary").alias("total_salary"),
    )
    .orderBy(F.col("avg_salary").desc())
)

q2.show()
````

## Solution 3 — Joins

````python
# 1. Inner Join
inner_join = (
    employeesDF
    .join(salesDF, "emp_id", "inner")
    .select("name", "department", "product", "price", "quantity", "sale_date")
)
print("=== Inner Join ===")
inner_join.show()

# 2. Left Join + null filter (employees with no sales)
no_sales = (
    employeesDF
    .join(salesDF, ["emp_id"], "left")
    .filter(F.col("sale_id").isNull())
    .select("emp_id", "name", "department")
)
print("=== No Sales (Left + null filter) ===")
no_sales.show()

# 3. Anti Join
anti_join = (
    employeesDF
    .join(salesDF, ["emp_id"], "left_anti")
    .select("emp_id", "name", "department")
)
print("=== Anti Join ===")
anti_join.show()

# 4. Departments with no employees
dept_no_emp = (
    departmentsDF.alias("d")
    .join(
        employeesDF.alias("e"),
        F.col("d.dept_name") == F.col("e.department"),
        "left",
    )
    .filter(F.col("e.emp_id").isNull())
    .select(F.col("d.dept_name"), F.col("d.dept_head"))
)
print("=== Departments with no Employees ===")
dept_no_emp.show()
````

## Solution 4 — rank, dense_rank, row_number

````python
window_spec = Window.partitionBy("department").orderBy(F.col("salary").desc())

ranked = (
    employeesDF
    .withColumn("rank", F.rank().over(window_spec))
    .withColumn("dense_rank", F.dense_rank().over(window_spec))
    .withColumn("row_number", F.row_number().over(window_spec))
)

print("=== All Ranks ===")
ranked.select(
    "name", "department", "salary", "rank", "dense_rank", "row_number"
).show()

print("=== Top 2 per Department ===")
(
    ranked
    .filter(F.col("dense_rank") <= 2)
    .select("name", "department", "salary", "dense_rank")
    .show()
)
````

## Solution 5 — lag, lead, running total

````python
sales_with_date = salesDF.withColumn("sale_date", F.to_date("sale_date"))

employee_window = Window.partitionBy("emp_id").orderBy("sale_date")
running_window = employee_window.rowsBetween(
    Window.unboundedPreceding, Window.currentRow
)

q5 = (
    sales_with_date
    .withColumn("prev_sale_price", F.lag("price", 1).over(employee_window))
    .withColumn("next_sale_price", F.lead("price", 1).over(employee_window))
    .withColumn("revenue", F.col("price") * F.col("quantity"))
    .withColumn("running_revenue", F.sum("revenue").over(running_window))
)

q5.select(
    "emp_id", "product", "price", "quantity", "sale_date",
    "prev_sale_price", "next_sale_price", "running_revenue",
).show()
````

## Solution 6 — when / otherwise

````python
q6 = (
    employeesDF
    .withColumn(
        "salary_band",
        F.when(F.col("salary").isNull(), "Unknown")
        .when(F.col("salary") >= 100000, "High")
        .when(F.col("salary") >= 70000, "Mid")
        .otherwise("Low"),
    )
    .withColumn(
        "experience_level",
        F.when(F.col("years_exp") >= 7, "Senior")
        .when(F.col("years_exp") >= 4, "Mid")
        .otherwise("Junior"),
    )
)

q6.select(
    "name", "salary", "salary_band", "years_exp", "experience_level"
).show()
````

## Solution 7 — Null Handling

````python
# 1. Count null salaries
null_count = employeesDF.filter(F.col("salary").isNull()).count()
print(f"Null salary count: {null_count}")

# 2. Fill null with the average salary
avg_salary = employeesDF.agg(F.avg("salary")).first()[0]
filled = employeesDF.na.fill({"salary": avg_salary})
print("=== Filled with avg ===")
filled.select("name", "salary").show()

# 3. coalesce
with_coalesce = employeesDF.withColumn(
    "salary", F.coalesce(F.col("salary"), F.lit(60000.0))
)
print("=== Coalesce ===")
with_coalesce.select("name", "salary").show()

# 4. Drop rows where any column is null
drop_all = employeesDF.na.drop()
print(f"After na.drop (any): {drop_all.count()} rows")

# 5. Drop rows only when both salary and country are null
drop_subset = employeesDF.na.drop(how="all", subset=["salary", "country"])
print(f"After subset na.drop (all): {drop_subset.count()} rows")
````

## Solution 19 — Complex Pipeline

````python
dept_window = Window.partitionBy("department").orderBy(
    F.col("salary").desc_nulls_last()
)

# Revenue per employee
revenue_df = (
    salesDF
    .withColumn("revenue", F.col("price") * F.col("quantity"))
    .groupBy("emp_id")
    .agg(F.round(F.sum("revenue"), 2).alias("revenue_per_emp"))
)

pipeline = (
    employeesDF
    .join(revenue_df, ["emp_id"], "left")
    .na.fill({"revenue_per_emp": 0.0})
    .join(
        departmentsDF,
        employeesDF["department"] == departmentsDF["dept_name"],
        "left",
    )
    .withColumn(
        "salary_pct_of_budget",
        F.round((F.col("salary") / F.col("budget")) * 100, 2),
    )
    .withColumn("dept_salary_rank", F.rank().over(dept_window))
    .withColumn(
        "perf_label",
        F.when(F.col("revenue_per_emp") > 3000, "Star")
        .when(F.col("revenue_per_emp") > 1000, "Good")
        .otherwise("Low"),
    )
    .select(
        "name", "department", "salary", "revenue_per_emp",
        "salary_pct_of_budget", "dept_salary_rank", "perf_label",
    )
    .orderBy("department", "dept_salary_rank")
)

pipeline.show()
````

## Solution 20 — Spark SQL

````python
# 1. Department with the highest average salary
print("=== Highest Avg Salary Dept ===")
spark.sql("""
  SELECT department, ROUND(AVG(salary), 2) AS avg_salary
  FROM employees
  GROUP BY department
  ORDER BY avg_salary DESC
  LIMIT 1
""").show()

# 2. Employees earning more than their department average
print("=== Above Dept Average ===")
spark.sql("""
  SELECT e.name, e.department, e.salary, ROUND(d.avg_sal, 2) AS dept_avg
  FROM employees e
  JOIN (
    SELECT department, AVG(salary) AS avg_sal
    FROM employees
    GROUP BY department
  ) d ON e.department = d.department
  WHERE e.salary > d.avg_sal
""").show()

# 3. Top two earners per department
print("=== Top 2 per Dept (SQL Window) ===")
spark.sql("""
  SELECT name, department, salary, rn FROM (
    SELECT name, department, salary,
           ROW_NUMBER() OVER (
             PARTITION BY department ORDER BY salary DESC
           ) AS rn
    FROM employees
  ) ranked
  WHERE rn <= 2
""").show()

# 4. Total revenue per department using all three tables
print("=== Revenue per Dept ===")
spark.sql("""
  SELECT e.department, d.dept_head,
         ROUND(SUM(s.price * s.quantity), 2) AS total_revenue
  FROM employees e
  JOIN sales s ON e.emp_id = s.emp_id
  JOIN departments d ON e.department = d.dept_name
  GROUP BY e.department, d.dept_head
  ORDER BY total_revenue DESC
""").show()

# 5. CTE: departments where salary exceeds 50% of budget
print("=== Salary > 50% of Budget (CTE) ===")
spark.sql("""
  WITH dept_salary AS (
    SELECT e.department, SUM(e.salary) AS total_salary, d.budget
    FROM employees e
    JOIN departments d ON e.department = d.dept_name
    GROUP BY e.department, d.budget
  )
  SELECT department,
         ROUND(total_salary, 0) AS total_salary,
         budget,
         ROUND(total_salary / budget * 100, 1) AS pct_of_budget
  FROM dept_salary
  WHERE total_salary > 0.5 * budget
  ORDER BY pct_of_budget DESC
""").show()
````
